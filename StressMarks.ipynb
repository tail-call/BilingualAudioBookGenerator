```python
from ruaccent import RUAccent

accentizer = RUAccent()
accentizer.load(omograph_model_size='turbo3.1', use_dictionary=True, tiny_mode=False)
```

```python
text = 'Это так же просто, как 2+2=4'
print(accentizer.process_all(text))
```

```python
def replace_accents(text):
    output = []
    characters = list(text)
    i = 0
    while i < len(characters):
        if characters[i] == "+":
            i += 1
            output.append(characters[i])
            output.append("\u0301")
        else:
            output.append(characters[i])
        i += 1
    return "".join(output)

replace_accents("чт+обы исп+ользовать формулир+овку б+изнес-пис+ателя +Эрика Б+ейкера")
```

```python
from wordfreq import word_frequency

babah = "чтобы использовать формулировку бизнес писателя Эрика Бейкера кушетки пределе трансформирован чертежей"

for word in babah.split():
    print(f"{word}: {word_frequency(word, "ru")}")

babah = "первая комната, в которую вы попадаете, выходя с улицы"

for word in babah.split():
    print(f"{word}: {word_frequency(word, "ru")}")
```

```python
for part in sentences:
    for subpart in part["sentences"]:
        if subpart['source'] == subpart["target"]:
            print(subpart)
```

```python
dictionary = {}

for part in sentences:
    for subpart in part["sentences"]:
        for word in subpart["uncommonWords"]:
            if word["source"] in dictionary:
                dictionary[word["source"]].append(word["target"])
            else:
                dictionary[word["source"]] = [word["target"]]

dictionary
```
