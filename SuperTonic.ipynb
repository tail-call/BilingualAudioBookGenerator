```python
from supertonic import TTS

tts = TTS(auto_download=True)
style = tts.get_voice_style(voice_name="F2")

text = """
Внутри\u0301 твое\u0301й реа\u0301льности гуля\u0301ют сквозняки\u0301. Внутри\u0301 твое\u0301й трево\u0301ги притаил\u0301ся партиза\u0301н. Внутри\u0301 твое\u0301й стери\u0301льности воня\u0301ет колбасо\u0301й. Я простуди\u0301лся, у\u0301мер, преврати\u0301лся в пластили\u0301н. Я иллюзо\u0301рен со всех сторон.
"""
wav, duration = tts.synthesize(text, voice_style=style, lang="ru")

tts.save_audio(wav, "supertonic.wav")
print(f"Generated {duration}s of audio")
```
