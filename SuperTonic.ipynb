```python
from supertonic import TTS

tts = TTS(auto_download=True)
style = tts.get_voice_style(voice_name="F2")

text = """
что́бы испо́льзовать формулиро́вку би́знес-писа́теля Э́рика Бе́йкера
"""
wav, duration = tts.synthesize(text, voice_style=style, lang="ru")

tts.save_audio(wav, "supertonic.wav")
print(f"Generated {duration}s of audio")
```
